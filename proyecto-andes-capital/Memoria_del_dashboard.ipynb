# 📊 Proyecto: Análisis Comercial Inmobiliario

El sector inmobiliario necesita evaluar su **desempeño comercial** para comprender su crecimiento, rentabilidad y comportamiento de clientes.

Se solicita construir un **dashboard ejecutivo** que permita analizar ventas, clientes y propiedades para apoyar decisiones estratégicas basadas en datos.


# 📂 Descripción de los Datasets

🔹 **hecho_ventas_propiedades**  
Tabla de hechos con las transacciones de venta (precio, cliente, propiedad, canal, fecha).

🔹 **dim_clientes**  
Información y segmentación de clientes.

🔹 **dim_propiedades**  
Características de las propiedades (tipo, tamaño, ubicación, etc).

🔹 **dim_fecha**  
Tabla calendario que deberá ser creada como parte del modelado.

# 🎯 Preguntas que el dashboard debe responder

**Desempeño general**

- ¿Cuál es el **ingreso total** generado por la empresa?
- ¿Cuántas **propiedades se han vendido**?
- ¿Cuál es el **precio promedio de venta**?
- ¿Cuál es la **comisión total** generada por la empresa?

**Análisis comercial**

- ¿Qué **tipo de propiedad genera más ingresos**?
- ¿Qué **segmento de clientes compra más propiedades**?
- ¿Qué canal de venta **genera mayor ingresos**?


**Análisis temporal**

- ¿Cómo evolucionan las **ventas en el tiempo**?
- ¿El negocio está **creciendo año contra año**?
- ¿Cómo va el desempeño **acumulado del año actual (YTD)**?

**Cohortes de clientes**

- ¿Los clientes **vuelven a comprar** después de su primera compra?
- ¿Qué **cohortes de clientes generan más compras/ingresos con el tiempo**?

## 📊 Paso 4: Creación de medidas

En este paso deberás crear las **medidas necesarias para responder las preguntas de negocio del dashboard**.

Las medidas pueden implementarse en **Power BI (DAX)** o como **Calculated Fields en Tableau**.

---

### 4.1 🔹 Medidas base

Crear como mínimo las siguientes medidas:

- **Ingreso Total** (suma del `precio_venta`)
- **Cantidad de Ventas**
- **Ticket Promedio**
- **Comisión Total**

Estas métricas permitirán analizar el desempeño general del negocio.

---

### 4.2 🔹 Medidas con contexto de filtro

Debes crear **al menos 2 medidas** que modifiquen el contexto de filtro.

En este caso se utilizarán **medidas de participación (%)**, que permiten entender **qué proporción del ingreso total aporta cada categoría**.

Las medidas que deben construirse son:

- **Participación de ingresos por tipo de propiedad**
- **Participación de ingresos por canal de venta**
- **Participación de ingresos por segmento de cliente**

Para ello:
- Modifica el contexto de filtro con las funciones propias de cada herramienta
  - En Power BI, usa `CALCULATE`
  - En Tableau, usa expresiones como `FIXED` (LOD)

Elimina el filtro de la dimensión correspondiente para obtener el total

Asegúrate de que el denominador represente el total sin el filtro de la categoría, mientras que el numerador sí lo conserva.

Estas medidas se utilizarán principalmente como **tooltip o etiqueta adicional en los gráficos** del dashboard para complementar el análisis.

---

### 4.3 🔹 Inteligencia de tiempo

Utilizando la tabla **`dim_fecha`** (Power BI) o las funciones de fecha en Tableau, debes crear al menos 2 de las siguientes medidas:

- **Ventas Year to Date (YTD)**
- **Ventas Month to Date (MTD)**
- **Ventas del año anterior**
- **Crecimiento Year over Year (YoY)**

Estas métricas permitirán analizar la evolución del negocio en el tiempo.

---

### 4.4 🔹 Medidas para análisis de cohortes

Debes crear cálculos que permitan analizar **si los clientes vuelven a comprar después de su primera compra**.

⚠️ Estos cálculos deben crearse como **columnas calculadas**, no como medidas.  
Todas deben crearse en la tabla **`hecho_ventas_propiedades`**.

Debes crear como mínimo:

- **Primera compra por cliente** → fecha mínima de compra de cada cliente.
- **Mes Cohorte** → mes en que el cliente realizó su **primera compra** (mes de adquisición).
- **Mes Venta** → mes en que ocurre cada transacción. Este campo permite identificar **cuándo el cliente vuelve a comprar** después de su primera compra.

Estos cálculos permitirán construir una **matriz de cohortes** donde:

- **Filas** representen el **mes de cohorte**
- **Columnas** representen el **mes de venta**
- **Valores** muestren **cantidad de ventas o ingresos**

## Resumen Ejecutivo

### Hallazgos clave

- Durante 2023 y 2024, Andes Capital generó 6.012,50 millones de unidades monetarias mediante 8.500 unidades vendidas. El ticket promedio fue de 707,35 mil unidades monetarias y las comisiones totales alcanzaron 200,63 millones de unidades monetarias.

- Las casas lideraron los ingresos con 2.240,54 millones de unidades monetarias. Este importe representa el 37,26 % del ingreso total. Se vendieron 2.324 unidades y el ticket promedio fue de 964,09 mil unidades monetarias.

- Las propiedades comerciales generaron 1.924,69 millones de unidades monetarias. Este importe representa el 32,01 % del ingreso total. Se vendieron 1.071 unidades y el ticket promedio fue de 1.797,10 mil unidades monetarias, el más alto de las tres categorías.

- Los departamentos generaron 1.847,28 millones de unidades monetarias. Este importe representa el 30,72 % del ingreso total. Se vendieron 5.105 unidades, equivalentes al 60,06 % de todas las operaciones. El ticket promedio fue de 361,86 mil unidades monetarias.

- Ciudad de México lideró los ingresos con 3.242,23 millones de unidades monetarias. Este importe representa el 53,92 % del ingreso total. En esta ciudad se vendieron 4.123 unidades y el ticket promedio fue de 786,38 mil unidades monetarias.

- Bogotá generó 2.770,27 millones de unidades monetarias. Este importe representa el 46,08 % del ingreso total. Bogotá lideró el volumen comercial con 4.377 unidades vendidas, equivalentes al 51,49 % de todas las operaciones. El ticket promedio fue de 632,92 mil unidades monetarias.

- El canal Corredor concentró aproximadamente el 72,85 % de los ingresos. El canal Directo generó el 27,15 % restante.

- El segmento Primera vez aportó 3.783,58 millones de unidades monetarias. Este importe representa el 62,93 % de los ingresos. El segmento registró 5.302 unidades vendidas y un ticket promedio de 713,61 mil unidades monetarias.

- El segmento Inversionista generó 1.471,44 millones de unidades monetarias. Este importe representa el 24,47 % del ingreso total. El segmento registró 2.093 unidades vendidas y un ticket promedio de 703,03 mil unidades monetarias.

- El segmento Alto patrimonio aportó 757,48 millones de unidades monetarias. Este importe representa el 12,60 % de los ingresos. El segmento registró 1.105 unidades vendidas y un ticket promedio de 685,50 mil unidades monetarias.

### Métricas principales

- Ingreso total: 6.012,50 millones de unidades monetarias

- Cantidad de ventas: 8.500 unidades

- Ticket promedio: 707,35 mil unidades monetarias

- Comisión total: 200,63 millones de unidades monetarias

- Crecimiento interanual de 2024 frente a 2023: 11,14 %

### Insights accionables

- Las casas constituyen el producto de mayor aportación económica. Generaron 2.240,54 millones de unidades monetarias, equivalentes al 37,26 % del ingreso total. Sin embargo, representaron solamente el 27,34 % de las ventas. Su ticket promedio de 964,09 mil unidades monetarias permite generar ingresos elevados con un volumen moderado de operaciones.

- Las propiedades comerciales aportaron 1.924,69 millones de unidades monetarias, correspondientes al 32,01 % del ingreso total. Esta categoría registró solamente 1.071 unidades vendidas. Su ticket promedio de 1.797,10 mil unidades monetarias fue el más elevado. Por ello, existe una oportunidad para aumentar los ingresos incrementando de forma controlada su volumen de ventas.

- Los departamentos concentraron 5.105 unidades vendidas, equivalentes al 60,06 % de todas las operaciones. Sin embargo, generaron el 30,72 % de los ingresos debido a su ticket promedio de 361,86 mil unidades monetarias. Esta categoría es el principal impulsor del volumen, pero no del valor por transacción.

- Ciudad de México generó 471,96 millones de unidades monetarias más que Bogotá, aunque vendió 254 unidades menos. Esta diferencia se explica porque el ticket promedio de Ciudad de México fue de 786,38 mil unidades monetarias, mientras que el de Bogotá fue de 632,92 mil unidades monetarias. La diferencia entre ambos tickets fue de 153,46 mil unidades monetarias.

- Bogotá lideró el volumen con 4.377 unidades vendidas, equivalentes al 51,49 % de las ventas. Ciudad de México lideró los ingresos con el 53,92 % del total. Por ello, Bogotá destaca por volumen y Ciudad de México por valor económico.

- El segmento Primera vez concentró 3.783,58 millones de unidades monetarias, equivalentes al 62,93 % de los ingresos. También reunió 5.302 unidades vendidas, correspondientes al 62,38 % de las ventas. Esto demuestra una elevada capacidad de captación, pero también una fuerte dependencia de nuevos compradores.

- Las cohortes de marzo y abril de 2023 presentaron la mayor recurrencia absoluta. La cohorte de marzo comenzó con 373 unidades vendidas y posteriormente alcanzó hasta 66 ventas recurrentes mensuales. La cohorte de abril comenzó con 365 unidades vendidas y alcanzó hasta 57 ventas recurrentes mensuales.

- Aunque existen recompras, la actividad posterior es considerablemente menor que la registrada en el mes inicial. Esto indica que la captación es sólida, pero existe margen para mejorar la retención de clientes.

- Los ingresos de 2024 aumentaron un 11,14 % frente a 2023. Este resultado refleja una evolución comercial positiva.

- El gráfico temporal incorpora líneas de ingreso mínimo, máximo y promedio para facilitar la comparación del desempeño mensual. La línea de tendencia presenta una trayectoria general ascendente durante el periodo analizado. Estas referencias permiten identificar con mayor claridad los meses que se encuentran por encima o por debajo del comportamiento promedio.

- El pronóstico estima ingresos de 190,43 millones de unidades monetarias para enero de 2025.

- Para febrero de 2025 se estiman ingresos de 205,30 millones de unidades monetarias.

- Para marzo de 2025 se proyectan ingresos de 347,68 millones de unidades monetarias. Esta evolución sugiere una recuperación moderada en febrero y un repunte considerablemente más fuerte en marzo.

- El pronóstico se calculó utilizando un intervalo de confianza del 95 %. Los límites del intervalo son amplios, por lo que las cifras previstas deben interpretarse como estimaciones orientativas y no como resultados garantizados.

### Recomendaciones estratégicas

- Priorizar la comercialización de casas, que lideran los ingresos con 2.240,54 millones de unidades monetarias, sin descuidar el volumen aportado por los departamentos.

- Incrementar de forma selectiva las ventas de propiedades comerciales. Esta estrategia permitiría aprovechar su ticket promedio de 1.797,10 mil unidades monetarias, el más alto de las tres categorías.

- Mantener una estrategia diferenciada por ciudad. En Ciudad de México se debe conservar el posicionamiento de valor y aumentar el ticket promedio. En Bogotá se debe trabajar en elevar el valor promedio de las operaciones, manteniendo su liderazgo en volumen.

- Fortalecer el canal Corredor, responsable de aproximadamente el 72,85 % de los ingresos. También se recomienda identificar las prácticas comerciales que explican su rendimiento.

- Implementar campañas de fidelización para el segmento Primera vez. El objetivo debe ser convertir parte de sus 5.302 unidades vendidas en compras recurrentes.

- Analizar las acciones aplicadas a las cohortes de marzo y abril de 2023. Se recomienda replicar en las cohortes recientes los factores que favorecieron su mayor recurrencia.

- Preparar acciones comerciales para marzo de 2025. El modelo proyecta para ese mes el mayor ingreso del trimestre, con una estimación de 347,68 millones de unidades monetarias.

- Actualizar el pronóstico cuando se incorporen nuevos meses de información. Disponer de más datos históricos permitirá reducir la incertidumbre de las estimaciones.
